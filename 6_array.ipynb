Aktivitas 6A

Tebak dulu, baru jalankan

Perhatikan tabel di atas (suhu = [28, 31, 30, 33, 29]). Sebelum menjalankan blok di bawah, tuliskan apa yang akan ditampilkan empat baris print terakhir. Lalu jalankan dan bandingkan. Mana yang membuat Anda terkejut?

```python
suhu = [28, 31, 30, 33, 29]       # suhu (°C) pada 5 hari

print("suhu[0] =", suhu[0])       # elemen pertama
print("suhu[4] =", suhu[4])       # elemen terakhir (len - 1)
print("suhu[-1] =", suhu[-1])     # indeks negatif: dihitung dari belakang

# --- tebak dulu, baru jalankan ---
print(suhu[1])
print(suhu[-2])
print(suhu[len(suhu) - 1])
print(suhu[2] + suhu[3])
```

suhu[-2] menghasilkan 33, bukan 31. Indeks negatif mulai dari -1 (bukan -0), jadi -2 adalah elemen kedua dari belakang, yaitu indeks 3.

suhu[2] + suhu[3] menghasilkan 63, bukan "3033". Karena isinya bilangan bulat, Python menjumlahkannya.

suhu[len(suhu) - 1] sama dengan suhu[-1]: keduanya mengambil elemen terakhir (29).

harga = [12, 15, 9] punya 3 elemen. Seorang mahasiswa ingin harga terakhir dan menulis harga[len(harga)]. Tebak apa yang terjadi, lalu jalankan di dalam try/except (ubah blok di atas bila perlu) dan tuliskan baris yang benar.

```python
harga = [12, 15, 9]
print("indeks yang sah: 0 sampai", len(harga) - 1)

try:
    print(harga[len(harga)])        # bug: len(harga) = 3, indeks 3 tidak ada
except IndexError as error:
    print("IndexError:", error)

# Perbaikan: elemen terakhir
print("terakhir:", harga[len(harga) - 1], "atau cukup", harga[-1])

# List kosong tidak punya indeks yang sah sama sekali
kosong = []
print("list kosong punya", len(kosong), "elemen")
```

harga = [12, 15, 9] punya 3 elemen dengan indeks yang sah 0, 1, dan 2. Kode mahasiswa harga[len(harga)] menghasilkan IndexError karena len(harga) bernilai 3, padahal indeks 3 tidak ada. Kesalahannya adalah mengira panjang list sama dengan indeks terakhir, padahal indeks dimulai dari 0 sehingga indeks terakhir adalah len(harga) - 1. Dengan try/except, error tersebut ditangkap dan program tidak berhenti, sehingga pesan list index out of range tampil sebagai output. Baris yang benar adalah harga[len(harga) - 1] atau lebih ringkas harga[-1], keduanya menghasilkan 9.

Aktivitas 6F

Tebak dulu, baru jalankan

Lanjutkan tabel secara manual untuk dua elemen terakhir (75 dan 70). Berapa total akhir dan rata-ratanya? Berapa skor yang bernilai 75 atau lebih? Lalu jalankan blok berikut.

```python
skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

Total dihitung dengan menjumlahkan tiap skor secara berurutan sehingga setelah dua elemen terakhir (75 dan 70) totalnya menjadi 310 lalu 380. Rata-ratanya adalah 380 dibagi 5 elemen, yaitu 76.0. Skor yang bernilai 75 atau lebih ada tiga, yaitu 80, 90, dan 75, sehingga jumlah_lulus bernilai 3.

Aktivitas 6G

Ganti ukuran jendela

Ubah blok di bawah menjadi rata-rata bergerak 5 hari (mulai perulangan di indeks 4, slice suhu[i - 4 : i + 1], bagi dengan 5). Berapa rata-rata yang Anda dapat dari 7 nilai? Mengapa?

Tetangga: perubahan harian dan rata-rata bergerak
Banyak pertanyaan membandingkan sebuah elemen dengan elemen sebelumnya (berapa perubahannya?) atau dengan tetangganya (berapa rata-rata 3 hari terakhir?). Untuk itu kita butuh indeks i. Berikut suhu 7 hari:

indeks	0	1	2	3	4	5	6

suhu	28	31	30	33	29	27	32

Perubahan harian: hari ini dikurangi kemarin. Hari ke-1 tidak punya kemarin, jadi perulangan dimulai dari indeks 1.

Rata-rata bergerak: rata-rata sebuah jendela yang bergeser satu langkah setiap kali. Jendela 3 hari yang berakhir di indeks 2 mencakup indeks 0, 1, 2; berikutnya mencakup 1, 2, 3; dan seterusnya. 7 nilai menghasilkan 7 - 3 + 1 = 5 rata-rata.

```python
suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 3 hari: jendela = 3 nilai yang berakhir di indeks i
rata_3 = []
for i in range(2, len(suhu)):
    jendela = suhu[i - 2 : i + 1]
    rata_3.append(round(sum(jendela) / 3, 1))

print("Rata-rata 3 hari:", rata_3)
print(len(suhu), "nilai ->", len(rata_3), "rata-rata")
```

Dengan jendela 5 hari, perulangan dimulai dari indeks 4 karena itu indeks pertama yang punya 5 nilai (indeks 0 sampai 4) dalam jendelanya. Dari 7 nilai, jendela bergeser sebanyak 7 − 5 + 1 = 3 kali, sehingga diperoleh 3 rata-rata, yaitu 30.2, 30.0, dan 30.2. Semakin besar ukuran jendela, semakin sedikit rata-rata yang dihasilkan, tetapi hasilnya semakin halus.
