Soal 6. Status Operasi Boiler

Program membaca suhu dan tekanan dari input, lalu mengubahnyaa menjadi float, Kondisi diperiksa dengan if-elif-else secara berurutan dari yang paling kriti s. Pertama, jika suhu mencapai 550 atau tekanan mencapai 14 (operator or, cukup salah satu), program mencetak TRIP. Jika tidak, program memeriksa apakah suhu lebih dari 520 dan tekanan lebih dari 12 (operator and, harus keduanya), lalu mencetak WARNING. Jika kedua kondisi tersebut tidak terpenuhi, program mencetak NORMAL. TRIP harus diperiksa lebih dulu agar kasus seperti suhu 540 dan tekanan 14 tidak salah terbaca sebagai WARNING.

```python
#Nama:Muhammad Taufik Furqan Al Aziz
#NIM:4232501002
#Tanggal:09/11/26
#Pemrograman Dasar

#8
volume = 100

while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")
    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")
```

Soal 8. Pengosongan Tangki Bahan Bakar

Program dimulai dengan volume = 100. Selama volume masih lebih dari 0, loop while berjalan: volume dikurangi 20 liter, volume terbaru dicetak, lalu program memeriksa apakah volume tepat 20 liter. Jika ya, pesan WARNING dicetak setelah baris volume. Loop berulang sebanyak 5 kali (80, 60, 40, 20, 0). Ketika volume mencapai 0, syarat volume > 0 menjadi salah sehingga loop berhenti, dan pesan tangki kosong dicetak satu kali karena posisinya berada di luar loop.

```python
#Nama:Muhammad Taufik Furqan Al Aziz
#NIM:4232501002
#Tanggal:09/11/26
#Pemrograman Dasar

#10
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for n, teks in enumerate(data, start=1):
    try:
        suhu = float(teks)
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break
    jumlah_valid += 1
    if suhu > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")

print(f"Jumlah data valid: {jumlah_valid}")
```

Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak

Program memakai for dengan enumerate(data, start=1) agar nomor titik dimulai dari 1. Pada tiap data, program mencoba mengubah teks menjadi float di dalam try. Jika berhasil, jumlah data valid ditambah 1, lalu suhu dibandingkan dengan 70.0 dan program mencetak Warning bila lebih tinggi (titik 3 bernilai 81.3). Jika gagal karena ValueError (titik 4 berisi “N/A”), program mencetak pesan data tidak valid lalu menjalankan break untuk menghentikan loop, sehingga titik 5 tidak diperiksa. Setelah loop selesai, program mencetak jumlah data valid, yaitu 3.
