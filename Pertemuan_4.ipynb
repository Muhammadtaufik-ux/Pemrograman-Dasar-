```python
#Latihan 1
tahun = int(input("Masukkan tahun: "))

if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
    print(tahun, "adalah tahun kabisat True")
else:
    print(tahun, "bukan tahun kabisat False")
```

```python
# Latihan 2
# Kalkulator Sederhana
# Nama : Muhammad Taufik Furqan Al Aziz | NIM : 4232501002

PEMBULATAN = 2          # variabel global (lingkup seluruh program)

def input_data():       # FUNGSI INPUT
    a = float(input("Masukkan angka pertama: "))   # variabel lokal
    b = float(input("Masukkan angka kedua  : "))   # variabel lokal
    return a, b

def hitung(a, b):       # OPERASI ARITMATIKA
    return {
        "Penjumlahan (+)": a + b,
        "Pengurangan (-)": a - b,
        "Perkalian (*)": a * b,
        "Pembagian (/)": a / b if b != 0 else "tidak terdefinisi",
        "Pembagian bulat (//)": a // b if b != 0 else "tidak terdefinisi",
        "Sisa bagi (%)": a % b if b != 0 else "tidak terdefinisi",
        "Pangkat (**)": a ** b,
    }

def output_data(a, b, hasil):   # FUNGSI OUTPUT
    print(f"\nHasil operasi {a} dan {b}:")
    for nama, nilai in hasil.items():
        if isinstance(nilai, float):
            nilai = round(nilai, PEMBULATAN)
        print(f"{nama:<22}: {nilai}")

# Program utama
x, y = input_data()
output_data(x, y, hitung(x, y))
```
