```python
#Muhammad Taufik Furqan Al Aziz, NIM : 4232501002
#26/09/2026
# Operasi aritmatika dua bilangan bulat
#Soal 1
bilangan1 = int(input("Masukkan bilangan pertama: "))
bilangan2 = int(input("Masukkan bilangan kedua: "))

print("Hasil penjumlahan (+):", bilangan1 + bilangan2)
print("Hasil pengurangan (-):", bilangan1 - bilangan2)
print("Hasil perkalian (*):", bilangan1 * bilangan2)
print("Hasil pembagian (/):", bilangan1 / bilangan2)
print("Hasil pembagian bulat (//):", bilangan1 // bilangan2)
print("Hasil modulus (%):", bilangan1 % bilangan2)
```

```python
#Muhammad Taufik Furqan Al Aziz, NIM:4232501002
#26/09/2026
# Soal 3
umur = int(input("Masukkan umur: "))
punya_sim = input("Apakah punya SIM? (True/False): ") == "True"

# Tanda kurung digunakan untuk menentukan urutan evaluasi
boleh_mengemudi = (umur >= 17) and punya_sim

print("Boleh mengemudi sendiri:", boleh_mengemudi)
```

```python
#Muhammad Taufik Furqan Al Aziz,NIM : 4232501002
#26/09/2026
#Soal 4
#Efisiensi termal
energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal: {efisiensi:.2f}%")
```

```python
#Muhammad Taufik Furqan Al Aziz, NIM:4232501002
#26/09/2026
# Soal 5
# Menentukan apakah unit pembangkit perlu maintenance

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(f"Apakah unit perlu maintenance? {perlu_maintenance}")
```
