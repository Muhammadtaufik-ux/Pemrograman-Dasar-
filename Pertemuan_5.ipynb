```python
# Latihan 2
#Nama : Muhammad Taufik Furqan Al Aziz
#NIM  : 4232501002

jumlah_panel = 12
daya_panel = 450        # Wp
jam_sinar = 5           # jam/hari
efisiensi = 0.8
kebutuhan = 25          # kWh/hari
tarif = 1444.70         # Rp/kWh

# 1. Energi per hari (kWh)
energi_harian = jumlah_panel * daya_panel / 1000 * jam_sinar * efisiensi

# 2. Energi per bulan (30 hari) dan nilai rupiahnya
energi_bulanan = energi_harian * 30
nilai_rupiah = energi_bulanan * tarif

# 3. Kekurangan energi per hari
kekurangan = kebutuhan - energi_harian

# 4. Apakah PLTS mencukupi?
cukup = energi_harian >= kebutuhan

print("Energi per hari   :", round(energi_harian, 2), "kWh")
print("Energi per bulan  :", round(energi_bulanan, 2), "kWh")
print("Nilai rupiah      : Rp", round(nilai_rupiah, 2))
print("Kekurangan/hari   :", round(kekurangan, 2), "kWh")
print("PLTS cukup?       :", cukup)
```

```python
#Latihan 3
#Nama : Muhammad Taufik Furqan Al Aziz
#NIM  : 4232501002

daya = 200              # kW
jam = 8                 # jam/hari
konsumsi = 0.25         # liter/kWh
harga_solar = 13500     # Rp/liter
tangki = 1000           # liter
arus = 20               # A
hambatan = 0.5          # ohm

# 1. Energi per hari (kWh)
energi = daya * jam

# 2. Kebutuhan solar per hari dan biayanya
solar_harian = energi * konsumsi
biaya = solar_harian * harga_solar

# 3. Hari penuh tangki cukup dan sisa solar
hari_penuh = int(tangki // solar_harian)
sisa_solar = tangki % solar_harian

# 4. Rugi daya kabel
rugi_daya = arus ** 2 * hambatan

# 5. Apakah rugi kabel tergolong tinggi?
tinggi = rugi_daya > 150 and arus >= 20

print("Energi per hari   :", energi, "kWh")
print("Solar per hari    :", solar_harian, "liter")
print("Biaya solar/hari  : Rp", biaya)
print("Tangki cukup      :", hari_penuh, "hari")
print("Sisa solar        :", sisa_solar, "liter")
print("Rugi daya kabel   :", rugi_daya, "W")
print("Rugi tergolong tinggi?", tinggi)
```
